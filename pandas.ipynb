```python
def cng_dur(a):
    h=int(a)
    z=(a-h)*60
    m=int(z)
    s=round((z-m)*60)
    if s>60:
        m=(m+s//60)
        s=s%60
    if m>60:
        h=h+(m//60)
        m=m%60

    return(f"{h:02d}:{m:02d}:{s:02d}")     
```

```python
cng_dur(a)
```

```python
df["Duration_hours_time"] = df["Duration_hours"].apply(cng_dur)
```

```python
def cng_del(a):
    h=0
    m=int(a)
    z=(a-m)*60
    s=round(z)

    if s>60:
        m=m+(s//60)
        s=s%60
    if m>60:
        h=h+(m//60)
        m=m%60
        
    return(f"{h:02d}:{m:02d}:{s:02d}")
```
